## 一、基础创建与查看

### 1. 导入 pandas，用字典创建 DataFrame：

- name: Tom, Jerry, Mike
- score: 85, 92, 78
- gender: 男，女，男

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
})
df
```

### 2. 查看上述 DataFrame 的前 1 行、形状、列名、数据类型。

```python
# 前1行
print("前1行：")
print(df.head(1))
print()

# 形状
print("形状：", df.shape)
print()

# 列名
print("列名：", df.columns.tolist())
print()

# 数据类型
print("数据类型：")
print(df.dtypes)
```

### 3. 用列表创建 DataFrame：

- data = [[1, "A"], [2, "B"], [3, "C"]]
- 列名设为 id、tag

```python
data = [[1, "A"], [2, "B"], [3, "C"]]
df2 = pd.DataFrame(data, columns=['id', 'tag'])
df2
```

## 三、缺失值

### 1. 创建含缺失值的 DataFrame：

- data = {"x":[1, None, 3], "y":[None, 5, 6]}
- 检测缺失值并统计每列缺失数量。

```python
df_na = pd.DataFrame({"x": [1, None, 3], "y": [None, 5, 6]})
print("DataFrame：")
print(df_na)
print()

# 检测缺失值
print("缺失值检测：")
print(df_na.isnull())
print()

# 统计每列缺失数量
print("每列缺失数量：")
print(df_na.isnull().sum())
```

### 2. 用 0 填充缺失值；删除含缺失值的行。

```python
# 用 0 填充缺失值
print("用 0 填充缺失值：")
print(df_na.fillna(0))
print()

# 删除含缺失值的行
print("删除含缺失值的行：")
print(df_na.dropna())
```

## 五、排序与去重

### 1. 按 score 升序排序。

```python
df.sort_values('score', ascending=True)
```

### 2. 给 DataFrame 添加一行重复数据（如 Tom, 85, 男），去重。

```python
# 添加重复行
new_row = pd.DataFrame({'name': ['Tom'], 'score': [85], 'gender': ['男'], 'grade': ['良好']})
df_dup = pd.concat([df, new_row], ignore_index=True)
print("添加重复行后：")
print(df_dup)
print()

# 去重
df_unique = df_dup.drop_duplicates()
print("去重后：")
print(df_unique)
```

## 六、合并与拼接

### 1. 横向合并：

- df1 = pd.DataFrame({"id":[1,2], "name":["X","Y"]})
- df2 = pd.DataFrame({"id":[1,2], "age":[20,30]})
- 按 id 合并。

```python
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})

df_merged = pd.merge(df1, df2, on='id')
df_merged
```

### 2. 纵向拼接：

- df_a = pd.DataFrame({"val":[1,2]})
- df_b = pd.DataFrame({"val":[3,4]})
- 拼接后重置索引。

```python
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})

df_concat = pd.concat([df_a, df_b], ignore_index=True)
df_concat
```

## 七、时间处理

### 1. 创建时间列：["2025-01-01","2025-01-02","2025-01-03"]，转为 datetime 格式。

```python
df_time = pd.DataFrame({"date": ["2025-01-01", "2025-01-02", "2025-01-03"]})
df_time['date'] = pd.to_datetime(df_time['date'])
print(df_time)
print()
print("数据类型：")
print(df_time.dtypes)
```

### 2. 提取时间列的月份、星期。

```python
df_time['month'] = df_time['date'].dt.month
df_time['weekday'] = df_time['date'].dt.dayofweek  # 0=周一, 6=周日
df_time
```

## 八、高级操作

### 1. 对 score 列做标准化（(值 - 均值) / 标准差）。

```python
df['score_normalized'] = (df['score'] - df['score'].mean()) / df['score'].std()
df[['name', 'score', 'score_normalized']]
```

### 2. 用透视表：行是 gender，值是 score，聚合函数为中位数。

```python
pd.pivot_table(df, values='score', index='gender', aggfunc='median')
```
